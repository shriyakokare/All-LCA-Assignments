```python
import re

def check_alphanumeric_regex(text):
    pattern = r'^[a-zA-Z0-9]+$'
    if re.match(pattern, text):
        return True
    return False

user_input = input("Enter a string: ")
if check_alphanumeric_regex(user_input):
    print("Valid: The string contains only a-z, A-Z and 0-9.")
else:
    print("Invalid: The string contains special characters or spaces.")
```
