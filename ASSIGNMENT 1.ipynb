```python
#create list
list1=["mango","orange","plum"]
print("list items are=",list1)
#access items
print("output of access item is",list1[0])
#add item
list1.append("orange")
print("the output of add item=",list1)
#remove item
list1.remove("plum")
print("the output of remove item=",list1)
#update item
list1[1]="blackcurrant"
print("The output of update item is",list1)
```

```python
#create tupple
tupple1=("kiwi","mango","avacado","banana","plum","apple","berry")
print("tupple items are",tupple1)
#access item
print("the output of access item is",tupple1[3])
#update tupple
y=list(tupple1)
#add item
y.append("orange")
tupple1=tuple(y)
print("The output of add item is",tupple1)
#remove item
y=list(tupple1)
y.remove("apple")
tupple1=tuple(y)
print("The output of remove item is",tupple1)
#update item
y=list(tupple1)
y[0]="kiwi"
tupple1=tuple(y)
print("The output of update item is",tupple1)
```

```python
#create dict
thisdict = {
    "brand" : "ford",
    "model" : "Mustang",
    "year" : 1964
}
print("dictionary is",thisdict)
#acess dict
x = thisdict.get("model")
print("The output of access dictionary is",x)
#add item
thisdict["color"] = "red"
print("The output of add dictionary is",thisdict)
#change item
thisdict["year"] = 2018
print("The output of change item is",thisdict)
#remove item
thisdict.pop("model")
print("The output of remove item is",thisdict)
```
